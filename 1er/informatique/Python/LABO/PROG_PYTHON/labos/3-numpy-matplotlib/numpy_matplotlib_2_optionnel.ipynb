```python
# Importation du module openCV
import cv2
```

```python
def get_mask(im):
    """Détecte un fond vert sur une image donnée

    Args:
        im (3D numpy array, uint8): Image sur fond vert, au format BGR

    Returns:
        2D numpy array, bool: masque de la partie de l'image à conserver (négatif du fond vert)
    """

    # Conversion en HSV (teinte-saturation-valeur)
    img_hsv = cv2.cvtColor(im, cv2.COLOR_BGR2HSV)
    h = img_hsv[:,:,0]
    s = img_hsv[:,:,1]
    #v = img_hsv[:,:,2]

    # Masque -> exclure ce qui est vert :
    #  H correspond à la teinte. Une valeur entre 30 et 80 correspond environ au vert
    #  S correspond à la saturation.
    #    Plus la saturation est élevée, plus on se rapproche d'une couleur "pure".
    #    Plus la saturation est faible, plus la couelur est proche du gris
    return np.logical_or(np.logical_or(h<30,h>80),s<50)
```
