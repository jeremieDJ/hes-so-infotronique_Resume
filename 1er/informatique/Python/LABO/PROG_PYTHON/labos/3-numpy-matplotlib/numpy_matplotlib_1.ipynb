## Exercice 2 : "Trouver le point d'intersection des deux droites en résolvant un système linéaire"

L'objectif est de poser un système linéaire pour trouver le point d'intersection de deux droites dans le plan et de le résoudre avec `np.linalg.solve`.

1. Trouver les équations des deux droites sous forme $z_1 = a_1 w_1 + b_1$, $z_2 = a_2 w_2 + b_2$.
2. On définit le vecteur inconnu $\bf x$ comme étant les coordonnées du point d'intersection des deux droites. Définir avec `np.array` la matrice $\bf A$ et le vecteur $\bf b$ du système linéaire $\bf A \bf x = \bf b$.
3. Résoudre le système linéaire en utilisant `np.linalg.solve`.
4. Extraire depuis le vecteur $\bf x$ les coordonnées du point d'intersection.
5. Marquer le point d'intersection dans le graphique avec une croix `x` noire selon le graphique ci-dessous. Indication : mettre comme 3ème argument de `plt.plot` la chaîne de caractère `'kx'`.

```python
### Votre code

w1 = np.array([1, 2, 3, 4])
w2 = np.array([2, 4, 6, 8])

# Vecteurs ordonnées
z1 = np.array([10, 20, 30, 40])
z2 = np.array([5, 15, 25, 35])

A = np.array([[-10,1],[-5,1]])

B = np.array([0,-5])

x = np.linalg.solve(A,B)
w_int, z_int = x  # coordonnées du point d'intersection

plt.plot(w_int, z_int, 'kx', markersize=10, label="Intersection")
plt.plot(w1,z1)   # afficher la parabole
plt.plot(w2,z2)   # afficher la parabole
plt.title("Intersection de deux droites")
plt.xlabel("abscisse x")
plt.ylabel("ordonnée y")
plt.grid()   # enclencher la grille


###
```

## Exercice 3 : "Polynômes et ses racines"

L'objectif est de saisir un polynôme de degré 3, calculer ses racines, et calculer les valeurs, où la pente de sa dérivée s'annulle.

1. Saisir avec `numpy.polynomial.Polynomial` le polynôme $p(x)=x^3-x^2-2x$ et printez le polynôme avec `print`.
2. Evaluer le polynôme $p(x)$ sur une grille de 100 points entre -2 et 3.
3. Afficher l'allure du polynôme sur un graphique.
4. Calculer et afficher les racines du polynôme $p(x)$ avec l'aide de la méthode `.roots()`.
5. Calculer la dérivée du polynôme $p(x)$ avec l'aide de la méthode `.deriv()`.
6. Calculer les abscisses correspondant à des minimums/maximums locaux. Indication : la dérivée s'annulle.
7. Afficher les différents points obtenus sur le graphique

```python
### Votre code
from numpy.polynomial import Polynomial

p = Polynomial([0, -2, -1, 1])  # p(x) = x^3 - x^2 - 2x
print(p)

x = np.linspace(-2, 3, 100)  # 100 points from -3 to 3
y = p(x)

y_racine = p.roots()
print(y_racine)

y_dervie = p.deriv()
print(y_dervie)

# Points où la dérivée s'annule (extrema)
extrema_x = y_dervie.roots()
extrema_y = p(extrema_x)

plt.plot(x,y)   # afficher la parabole
plt.plot(x,y_dervie(x))   # afficher la parabole
plt.title("Polynômes et ses racines")
plt.xlabel("abscisse x")
plt.ylabel("ordonnée y")
plt.grid()   # enclencher la grille

# Plot the roots as red points on the x-axis
plt.scatter(y_racine, np.zeros_like(y_racine), color='red', label="Racines")
plt.scatter(extrema_x, np.zeros_like(extrema_x), color='blue', label="deriver")
plt.scatter(extrema_x, extrema_y, color='green', label="Extrema")


plt.legend()
###
```
