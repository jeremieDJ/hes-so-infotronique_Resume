
# Lecture et écriture de fichiers : Théorie

## Fonction `open()`

En python, la fonction `open()` permet d'ouvrir un fichier. Elle prend 2 paramètres :

* Le nom du fichier, incluant le chemin d'accès. Le chemin d'accès peut être absolu ou relatif :
    * En absolu, où le chemin est spécifié depuis le répertoire racine (`C:` ou `D:` sur Windows, `/` sur Linux ou MacOS)
    * En relatif, depuis le répertoire où se trouve le programme exécuté. `.` peut être utilisé pour désigner le répertoire courant, et `..` pour le répertoire parent.
* Le mode d'ouverture, pouvant prendre 3 valeurs différentes :
    * `'r'` (read) pour la lecture.
    * `'w'` (write) pour l'écriture. Si le fichier n'existe pas, il est créé. Sinon, il est écrasé.
    * `'a'` (append) pour l'ajout. Si le fichier n'existe pas, il est créé. Sinon, le fichier existant est ouvert, et du contenu peut être ajouté à la suite du contenu existant.

En plus de ça, on peut séparer les fichiers en 2 types :

* les fichiers texte, qui stockent une chaîne de caractères. Ces fichiers peuvent être lus à l'aide d'un éditeur de texte.
* les fichiers binaires, où l'information stockée peut être vue comme une suite de bytes. Pour être lus, ces fichiers doivent en général être ouverts avec un programme spécialisé (par exemple MS Word pour un \*.docx, Excel pour un \*.xlsx, etc.).

Par défaut, python ouvre les fichiers en mode texte. Pour ouvrir un fichier binaire, vous devez ajouter `b` au paramètre spécifiant le mode d'ouverture (`'rb'` au lieu de `'r'`, `'wb'` au lieu de `'w'`, `'ab'` au lieu de `'a'`).

NB: sur windows, c'est le caratère `'\'` qui est utilisé comme séparateur de répertoire (p ex. `'.\dossier\sousDossier\fichier'`). Touefois, en python, `'\'` est utilisé pour spécifier des caractères spéciaux. Vous devrez donc utiliser une double-backslash (`'.\\dossier\\sousDossier\\fichier'`). L'alternative et de simplement utiliser un slash (`./dossier/sousDossier/fichier`). Cela permet d'alléger l'écriture, et a également l'avantage d'être compatible avec la notation des autres systèmes d'exploitation, qui utilisent le slash.

## Lecture et écriture de texte

Une fois le fichier ouvert on dispose de plusieurs fonctions pour lire ou écrire du texte :

* `read()`, qui lit l'intégralité du fichier et la retourne sous forme de chaîne de caractères.
* `readline()`, qui lit une seule ligne du fichier et la retourne sous forme de chaîne de caractères. En appelant la fonction pour la N\textsuperscript{ème} fois, on lit la N\textsuperscript{ème} ligne.
* `readlines()`, qui lit l'intégralité du fichier, et la retourne sous forme de liste, où chaque élément de la liste correspond à une ligne.
* `write()`, qui écrit une chaîne de caractère donnée.
* `writelines()`, qui permet de passer une liste de chaînes de caractères.

Ces fonctions sont des méthodes de l'objet retourné par la fonction `open()`.

## Fermeture d'un fichier

Une fois l'opération de lecture ou d'écriture effectuée, le fichier doit être refermé. Si on ne referme pas le fichier, on risque d'être confronté à certains problèmes, par exemple l'impossibilité d'ouvrir le fichier avec un autre programme. On a 2 possibilités :

* Il est possible d'utiliser la fonction `close()`. Dans ce cas, le code pour lire un fichier sera :

    >``` python
    >f = open('./exemple.txt','r')
    >file_content = f.read()
    >f.close()
    >```

* Au lieu de devoir refermer explicitement le fichier après la lecture, on peut utiliser le mot-clé `with`. Cette solution est plus sûre, car elle permet de s'assurer que python referme le fichier dans tous les cas, même si une exception est levée pendant que le fichier est ouvert. La syntaxe à utiliser est la suivante :

    >``` python
    >with open('./exemple.txt', 'r') as f:
    >    file_content = f.read()
    >```

    Dans ce cas, le fichier sera automatiquement refermé à la fin du bloc d'instruction qui suit la ligne `with open(...) as f:`.


### Exemples d'écriture et de lecture


````python

# Ecriture et lecture d'un fichier, sans l'utilisation du mot-clé 'with' :

f = open('./data/test1.txt', 'w')
f.write('Hello world !')
f.close()

f = open('./data/test1.txt', 'r')
print(f.read())
f.close()
````

````python

# Ecriture et lecture d'un fichier, avec l'utilisation du mot-clé 'with' :

with open('./data/test2.txt', 'w') as f:
    f.write('Hello world !')

with open('./data/test2.txt', 'r') as f:
    print(f.read())
````


## Les modules `os` et `shutil`

Les modules `os` et `shutil` peuvent être utiles dans certains cas, pour certaines opérations liées au système de fichiers, notamment :

* Pour la manipulation de chaînes de caractères représentant des chemins d'accès et des noms de fichiers, plusieurs fonctions du module `os.path` :
    * `os.path.join(path, *paths)` permet d'assembler plusieurs chaînes de caractère et retourne un chemin d'accès valide, en gérant l'ajout de `/` ou `\` entre les chaînes.
    * Les fonctions `os.path.basename()`, `os.path.split()`, `os.path.splitext()` et d'autres, permettant de séparer, depuis un chemin donné, le dossier, le nom de fichier, l'extension, etc.
* Pour lister des fichiers ou vérifier l'existence de fichiers :
    * `os.listdir(path)` retourne une liste contenant les noms des fichiers et répertoires dans le dossier `path`.
    * `os.path.exists(path)` retourne `True` s'il existe un fichier ou un répertoire au chemin donné.
    * `os.path.isdir(path)` retourne `True` s'il existe un dossier au chemin donné.
    * `os.path.isfile(path)` retourn `True` s'il existe un fichier au chemin donné.
* Pour copier, déplacer, renommer ou supprimer des fichiers :
    * `shutil.copy(...)`
    * `os.rename(...)`, qui permet à la fois de déplacer ou de renommer un fichier.
    * `os.remove(...)`
* Pour créer ou supprimer des répertoires :
    * `os.mkdir(...)`, `os.makedirs(...)`
    * `os.rmdir(...)`

Les détails concerant l'utilisation de ces fonctions se trouvent dans la documentation : 

* https://docs.python.org/3/library/os.html
* https://docs.python.org/3/library/os.path.html
* https://docs.python.org/3/library/shutil.html


````python

# Exemple d'utilisation du module os :

import os

print(os.listdir('.'))          # Liste des fichiers dans le répertoire courant
print(os.listdir('..'))         # Liste des fichiers dans le répertoire parent
print(os.listdir('./data'))     # Liste des fichiers dans le répertoire "intro"
print('')

print(os.path.isdir('./data'))              # Dossier -> doit retourner True
print(os.path.isdir('./fichiers.ipynb'))    # Fichier -> doit retourner False
print('')

print(os.path.exists('./data/test1.txt'))  # Fichier existant
print(os.path.exists('./data/test3.txt'))  # Fichier inexistant
print('')

````

En plus des fichiers textes simples, il existe plusieurs manières de formatter des fichiers, selon le type d'information qu'on souhaite stocker. Ce chapitre en liste quelques uns, que vous pourriez rencontrer ou qui pourraient vous être utile.

### CSV

CSV (Comma-Separated Values) est un type de fichier qui permet de stocker des tableaux de données, en 2 dimensions. Un séparateur permet de connaître les limites de chaque colonne du tableau. Ce séparateur est habituellement une virgule, mais peut être représenté par n'importe quel caractère, notamment une tabulation ou un point-virgule.

Ce format est très utilisé, par exemple pour exporter les données des appareils de mesure. Comme il s'agit de texte, il a l'avantage d'être lisible depuis un éditeur de texte. Par contre, il n'est pas optimal pour de grandes quantités de données, car il prend beaucoup plus de mémoire qu'un fichier binaire qui stockerait la même information.

En python, il existe plusieurs modules permettant de lire et écrire ce type de fichiers, par exemple `csv` et `pandas`. Mais il est possible, et généralement plus simple, d'utiliser directement des fonctions de la classe `str` afin de formatter le texte.

#### Exemple de données au format CSV

Voici un exemple de données au format CSV, représentant deux tensions mesurées en fonction du temps :

>```csv
>Temps;Canal A;Canal B
>(ms);(mV);(mV)
>

>-2.49933615;3.32800200;8.36614100

>```

La lecture de ce fichier peut être faite de la manière suivante :

### JSON / YAML

JSON et YAML sont des langages de balisage, permettant de sauver des données de types simples, pouvant être notamment des dictionnaires et des listes. Un fichier peut uniquement contenir des données des types suivants :

* Dictionnaires
* Listes
* Chaînes de caratères
* Nombres (entiers ou flottants)

Vous allez probablement rencontrer ces 2 types de fichiers, ayant une syntaxe légèrement différente. Dans les 2 cas, il existe des modules python permettant de lire et écrire ces fichiers.

#### Exemple de données au format JSON

>```json
>{
>    "Nom":"Doe",
>    "Prénom":"John",
>    "Age": 33,
>    "Enfants":[
>        "Alice",
>        "Bob",
>        "Charlie"
>    ]
>}
>```

#### Sauvegarde de données au format JSON

````python

d = {
    'Nom':'Doe',
    'Prénom':'John',
    'Age': 33,
    'Enfants':['Alice', 'Bob', 'Charlie'
    ]
}

import json

with open('./data/exemple.json', 'w', encoding='utf-8') as f:
    json.dump(d,f, indent=4, ensure_ascii=False)
````


#### Exemple de données au format YAML
>```yaml
>Nom: Doe
>Prénom: John
>Age: 33
>Enfants:
> - Alice
> - Bob
> - Charlie
>```

#### Sauvegarde de données au format YAML

NB : le paquet `pyyaml` doit d'abord être installé (`pip install pyyaml`) :


````python

import yaml

with open('data/exemple.yml', 'w') as f:
    yaml.dump(d,f, allow_unicode=True, sort_keys=False)
````

### Pickle

Pickle permet de sauver des objets python, par exemple des objets ndarray (numpy), en format binaire. Cela permet de sauver des données plus complexes que JSON et YAML, sans devoir préalablement les mettre en forme. On peut par exemple instancier un objet d'une classe qu'on a préalablement définie, et le sauvegarder dans un fichier pickle après avoir modifié ses attributs. Plus tard, on poura charger le fichier et obtenir une copie de l'objet sauvegardé.

Il est spécifique à python, ce qui signifie que le fichier enregistré ne pourra généralement pas être lu par un autre programme. De plus, si pickle est utilisé pour transférer des données entre 2 PC n'ayant pas le même environnement, il est possible que cela donne lieu à des erreurs, dues à des conflits de versions.

#### Sauvegarde de données au format Pickle



````python

import pickle

with open('data/exemple.pickle', 'wb') as f:
    pickle.dump(d,f)
````
