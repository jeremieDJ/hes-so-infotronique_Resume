```python
import numpy as np
import matplotlib.pyplot as plt
```

# Chargement des fichiers de mesure

on trouve dans le répertoire "acquisitions" les fichiers CSV qui contiennent des mesures pour un poids et la valeur de l'ADC

```python
from pathlib import Path

fichiers_csv = sorted(Path("acquisitions").glob("*.csv"))
donnees = [
    np.loadtxt(fichier, delimiter=",", skiprows=1)
    for fichier in fichiers_csv
]

print([fichier.name for fichier in fichiers_csv])
```

Fait un seul tableau à partir de toutes les mesures

```python
data_id = np.concatenate(donnees[5:10], axis=0)
data_full = np.concatenate(donnees, axis=0)
```

```python
plt.plot(data_id[:,   0], data_id[:, 1],'o')
plt.grid()
plt.xlabel("Poids (g)")
plt.ylabel("ADC (CODE)")
```

Converti le code ADC en mV

```python
data_id_mv=data_id*np.array([1, 3300/2**24]) # conversion en mV
data_full_mv=data_full*np.array([1, 3300/2**24]) # conversion en mV
```

```python
plt.plot(data_id_mv[:, 0], data_id_mv[:, 1],'o')
plt.grid()
plt.ylabel("Tension (mV)")
plt.xlabel("Poids (kg)")
```
